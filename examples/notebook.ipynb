# Display Caption and Adjust Widget Style

## Import Packages

```python
import ipywidgets as widgets

from pingi.notebook import adjust_notebook_widget_style, display_caption
```

## Display a Caption

```python
display_caption("A 16 px bold italic caption containing some LaTeX: $E=mc^2$.")
```

## Display a Widget Stylized with VS Code's Theme

### Adjust Widgets Styling

Note that `adjust_notebook_widget_style()` injects global CSS into the notebook, so that
the VS Code theme styling is applied to all existing and future Jupyter widgets in the
current session.

```python
adjust_notebook_widget_style()
```

### Display Widget

```python
widgets.VBox(
    [
        widgets.IntProgress(value=70, min=0, max=100, description="Progress:"),
        widgets.FloatSlider(value=0.3, min=0, max=1, description="Alpha:"),
        widgets.Dropdown(
            options=["Option A", "Option B", "Option C"],
            value="Option A",
            description="Choice:",
        ),
        widgets.Text(value="Sample text", description="Text:"),
        widgets.Button(description="Run", button_style="primary"),
    ]
)
```
